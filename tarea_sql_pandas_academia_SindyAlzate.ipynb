## 3. Cargar tablas a SQLite

### Punto 6. JOIN entre tablas

Muestra:
- nombre del estudiante
- nombre del curso
- estado de la inscripción

Usa `estudiantes`, `inscripciones` y `cursos`.

**Entregable:**
- SQL
- pandas

```python
# Punto 6 - SQL
query_6 = """
SELECT
    e.nombre,
    c.nombre_curso,
    i.estado
FROM estudiantes AS e
INNER JOIN inscripciones AS i
    ON e.estudiante_id = i.estudiante_id
INNER JOIN cursos AS c
    ON i.curso_id = c.curso_id
"""
sql_6 = run_sql(query_6)
sql_6
```

```python
# Punto 6 - pandas
resultado = (
    estudiantes
    .merge(inscripciones, on="estudiante_id", how="inner")
    .merge(cursos, on="curso_id", how="inner")
    [["nombre", "nombre_curso", "estado"]]
)

resultado
```

### Punto 8. HAVING

Calcula cuántas inscripciones tiene cada estudiante y muestra solo aquellos con **2 o más**.

Debes mostrar:
- `estudiante_id`
- `nombre`
- `n_inscripciones`

**Entregable:**
- SQL
- pandas

```python
# Punto 8 - SQL
query_8 = """
SELECT
    e.estudiante_id,
    e.nombre,
    COUNT(i.inscripcion_id) AS n_inscripciones
FROM estudiantes AS e
INNER JOIN inscripciones AS i
    ON e.estudiante_id = i.estudiante_id
GROUP BY
    e.estudiante_id,
    e.nombre
HAVING COUNT(i.inscripcion_id) >= 2
"""

sql_8 = run_sql(query_8)
sql_8
```

```python
# Punto 8 - pandas
inscripciones_por_estudiante = (
    inscripciones
    .groupby("estudiante_id")
    .size()
    .reset_index(name="n_inscripciones")
)

resultado = (
    estudiantes[["estudiante_id", "nombre"]]
    .merge(inscripciones_por_estudiante, on="estudiante_id", how="inner")
)

resultado = resultado[
    resultado["n_inscripciones"] >= 2
]

resultado
```

### Punto 9. CASE WHEN / segmentación

Clasifica los cursos según su precio:
- `Bajo` si el precio es menor a 120
- `Medio` si el precio está entre 120 y 180 inclusive
- `Alto` si el precio es mayor a 180

Debes mostrar:
- `nombre_curso`
- `precio`
- `segmento_precio`

**Entregable:**
- SQL
- pandas

```python
# Punto 9 - SQL
query_9 = """
SELECT
    nombre_curso,
    precio,
    CASE
        WHEN precio < 120 THEN 'Bajo'
        WHEN precio BETWEEN 120 AND 180 THEN 'Medio'
        WHEN precio > 180 THEN 'Alto'
    END AS segmento_precio
FROM cursos
"""

sql_9 = run_sql(query_9)
sql_9
```

```python
# Punto 9 - pandas
cursos_segmentados = cursos.copy()

cursos_segmentados["segmento_precio"] = pd.cut(
    cursos_segmentados["precio"],
    bins=[-float("inf"), 119.99, 180, float("inf")],
    labels=["Bajo", "Medio", "Alto"]
)

resultado = cursos_segmentados[
    ["nombre_curso", "precio", "segmento_precio"]
]

resultado
```

### Punto 10. Ejercicio final integrador

Calcula el total pagado por categoría de curso.

Debes:
- unir `cursos`, `inscripciones` y `pagos`
- agrupar por `categoria`
- sumar `monto`
- ordenar de mayor a menor

**Entregable:**
- SQL
- pandas

```python
# Punto 10 - SQL
query_10 = """
SELECT
    c.categoria,
    SUM(p.monto) AS total_pagado
FROM cursos AS c
INNER JOIN inscripciones AS i
    ON c.curso_id = i.curso_id
INNER JOIN pagos AS p
    ON i.inscripcion_id = p.inscripcion_id
GROUP BY c.categoria
ORDER BY total_pagado DESC
"""

sql_10 = run_sql(query_10)
sql_10
```

```python
# Punto 10 - pandas
resultado = (
    inscripciones
    .merge(cursos[["curso_id", "categoria"]], on="curso_id", how="inner")
    .merge(pagos[["inscripcion_id", "monto"]], on="inscripcion_id", how="inner")
    .groupby("categoria")["monto"]
    .sum()
    .sort_values(ascending=False)
    .reset_index(name="total_pagado")
)

resultado
```
